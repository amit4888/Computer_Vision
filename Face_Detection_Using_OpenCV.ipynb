```python
import cv2
import matplotlib.pyplot as plt

# Load the image in BGR format (default OpenCV)
img_bgr = cv2.imread(r'C:\Users\AMIT KUMAR\OneDrive\Pictures\Screenshot 2024-09-18 004211.png')

# Convert to RGB for display and face detection
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Convert to grayscale for detection
gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

# Load Haar Cascade
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

# Detect faces
faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5)

# Draw rectangle on img_rgb (not on grayscale)
for (x, y, w, h) in faces:
    cv2.rectangle(img_rgb, (x, y), (x + w, y + h), (255, 0, 0), 2)  # Blue rectangle in RGB

# Show using matplotlib
plt.imshow(img_rgb)
plt.title('Face Detection')
plt.show()
```
